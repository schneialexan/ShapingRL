```python
import matplotlib.pyplot as plt
import matplotlib.image as mpimg
import numpy as np
```

```python
GAME_IMAGE_FRAME_HEIGHT = 210
GAME_IMAGE_FRAME_WIDTH = 160
```

```python
n_grid_items_horizontal = 7
n_grid_items_vertical = 10

horizontal_spacing = int(GAME_IMAGE_FRAME_HEIGHT / n_grid_items_horizontal)
vertical_spacing = int(GAME_IMAGE_FRAME_WIDTH / n_grid_items_vertical)

# Path to your image file (update this to your own image path)
image_path = "../tests/information_extraction/frames/montezumas/0_lives.png"

# Read the image
img = mpimg.imread(image_path)

# Display the image
plt.imshow(img)
plt.title("Real Image with Raster Grid")

# Add raster (grid lines)
rows, cols = img.shape[0], img.shape[1]
plt.xticks(np.arange(0, cols, horizontal_spacing))  # Adjust grid spacing (e.g., 50 pixels)
plt.yticks(np.arange(0, rows, vertical_spacing))
plt.grid(True, color='white', linewidth=0.8)

# Optionally, make grid lines align exactly with pixels
plt.gca().set_xticks(np.arange(-0.5, cols, 50), minor=True)
plt.gca().set_yticks(np.arange(-0.5, rows, 50), minor=True)
plt.grid(which='minor', color='gray', linestyle='-', linewidth=0.5)

plt.show()
```
